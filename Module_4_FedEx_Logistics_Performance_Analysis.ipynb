```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
url = 'https://raw.githubusercontent.com/anujsharmads/FedEx-Logistics-Performance-Analysis/refs/heads/main/SCMS_Delivery_History_Dataset%20(1).csv'

df = pd.read_csv(url)
df.head(2)
```

```python
df.info()
```

```python
null_count = df.isnull().sum().sort_values(ascending = False)
null_count[null_count>0]
```

```python
df_copy = df.copy()
```

```python
df_copy['Shipment Mode']=df_copy['Shipment Mode'].fillna(df_copy['Shipment Mode'].mode()[0])
```

```python
df_copy['Line Item Insurance (USD)']=df_copy['Line Item Insurance (USD)'].fillna(df_copy['Line Item Insurance (USD)'].median(numeric_only=True))
```

```python
df_copy.drop(columns='Dosage', inplace = True)
```

```python
df_copy.isnull().sum().sum()
```

```python
df_copy.columns
```

```python
df_copy['Weight (Kilograms)']=pd.to_numeric(df_copy['Weight (Kilograms)'],errors = 'coerce')
```

```python
df_copy['Freight Cost (USD)']=pd.to_numeric(df_copy['Freight Cost (USD)'],errors = 'coerce')
```

```python
df_copy['Weight (Kilograms)']=df_copy['Weight (Kilograms)'].fillna(df_copy['Weight (Kilograms)'].median())
df_copy['Freight Cost (USD)']=df_copy['Freight Cost (USD)'].fillna(df_copy['Freight Cost (USD)'].median())
```

```python
df_copy['PQ First Sent to Client Date']=pd.to_datetime(df_copy['PQ First Sent to Client Date'],format = 'mixed', errors = 'coerce')
```

```python
df_copy['PO Sent to Vendor Date']=pd.to_datetime(df_copy['PO Sent to Vendor Date'],format = 'mixed', errors = 'coerce')
```

```python
df_copy['Scheduled Delivery Date']=pd.to_datetime(df_copy['Scheduled Delivery Date'],format = '%d-%b-%y', errors = 'coerce')
```

```python
df_copy['Delivered to Client Date']=pd.to_datetime(df_copy['Delivered to Client Date'],format = '%d-%b-%y', errors = 'coerce')
```

```python
df_copy['Delivery Recorded Date']=pd.to_datetime(df_copy['Delivery Recorded Date'],format = '%d-%b-%y', errors = 'coerce')
```

```python
print(df_copy['Weight (Kilograms)'].dtype)
print(df_copy['Freight Cost (USD)'].dtype)
print(df_copy['PQ First Sent to Client Date'].dtype)
print(df_copy['PO Sent to Vendor Date'].dtype)
print(df_copy['Scheduled Delivery Date'].dtype)
print(df_copy['Delivered to Client Date'].dtype)
print(df_copy['Delivery Recorded Date'].dtype)
```

```python
print(df_copy['Weight (Kilograms)'].isnull().sum())
print(df_copy['Freight Cost (USD)'].isnull().sum())
print(df_copy['PQ First Sent to Client Date'].isnull().sum())
print(df_copy['PO Sent to Vendor Date'].isnull().sum())
print(df_copy['Scheduled Delivery Date'].isnull().sum())
print(df_copy['Delivered to Client Date'].isnull().sum())
print(df_copy['Delivery Recorded Date'].isnull().sum())
```
